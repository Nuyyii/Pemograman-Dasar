```python
# Inisialisasi variabel dengan berbagai tipe data
angka_int = 10         # integer
angka_float = 2.5      # float
teks = "Mi ayam bakso"  # string
angka_hex = 255        # integer yang akan ditampilakan sebagai hexadesimal
angka_oct = 8          # integer yang akan ditampilakan sebagai oktal

# Operator %d (Integer)
print("Jumlah pesanan: %d porsi" % (angka_int))

# Operator %f (Float)
print("Harga total: Rp %f" % (angka_float))

# Operator %s (String)
print("Hendi sedang memesan %s" % (teks))

# Operator %x (Hexadesimal)
print("Nilai hexadesimal dari %d adalah %x" % (angka_hex, angka_hex))

# Operator %o (Octal)
print("Nilai oktal dari %d adalah %o" % (angka_oct, angka_oct))
```

```python
# Konversi Waktu

# Mengubah variabel total detik agar bisa menerima input
total_detik = int(input("Masukkan total detik: "))

# Menghitung jumlah jam dan sisa detik setelah diambil jamnya
jam = total_detik // 3600
sisa_detik = total_detik % 3600

# Menghitung jumlah menit dan detik akhir dari sisa detik
menit = sisa_detik // 60
detik = sisa_detik % 60

# Menampilkan hasil konversi
print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
# Modifikasi Program Menghitung Nilai Akhir

# Modifikasi Poin 2: Nilai mahasiswa terdiri dari 5 variabel input keyboard
tugas1 = float(input("Masukkan nilai Tugas 1 : "))
tugas2 = float(input("Masukkan nilai Tugas 2 : "))
kuis   = float(input("Masukkan nilai Kuis    : "))
uts    = float(input("Masukkan nilai UTS     : "))
uas    = float(input("Masukkan nilai UAS     : "))

# Menghitung nilai akhir mahasiswa (contoh bobot total 100%)
# Bobot: Tugas1 (15%), Tugas2 (15%), Kuis (10%), UTS (30%), UAS (30%)
nilai_mahasiswa = (0.15 * tugas1) + (0.15 * tugas2) + (0.10 * kuis) + (0.30 * uts) + (0.30 * uas)

# Modifikasi Poin 1: Input standar kelulusan melalui keyboard
standar_kelulusan = float(input("Masukkan standar kelulusan: "))

# Evaluasi logika kelulusan (Operator Perbandingan)
apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print("\n--- HASIL EVALUASI ---")
print(f"Nilai Mahasiswa: {nilai_mahasiswa:.2f}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
# Modifikasi 1: Input total belanja via keyboard
belanja = float(input("Masukkan total belanja Anda: Rp "))

# Modifikasi 2: Diskon 35% (0.35) jika belanja > 1000000
# Menggunakan teknik tuple indexing [belanja > 1000000] (0 = False, 1 = True)
diskon = (0, 0.35 * belanja)[belanja > 1000000]

# Menghitung total harga pembayaran
total_harga = belanja - diskon

# Menampilkan hasil
print("\n--- DETAIL PEMBAYARAN ---")
print(f"Total Belanja : Rp {belanja:,.2f}")
print(f"Potongan Diskon: Rp {diskon:,.2f}")
print(f"Total Bayar   : Rp {total_harga:,.2f}")
```

```python
# LATIHAN 1: MENENTUKAN TAHUN KABISAT

# Input tahun dari keyboard
tahun = int(input("Masukkan tahun: "))

# Ekspresi Logika Kompleks
is_kabisat = ((tahun % 4 == 0) and (tahun % 100 != 0)) or (tahun % 400 == 0)

# Output hasil
print(f"Apakah tahun {tahun} merupakan tahun kabisat? {is_kabisat}")
```

```python
# LATIHAN 2: PROGRAM KASIR TOKO BUKU SEDERHANA
def hitung_pembayaran():
    # Fungsi Input via Keyboard
    nama_pembeli = str(input("Masukkan nama pembeli: "))
    harga_buku = float(input("Masukkan harga per buku (Rp): "))
    jumlah_buku = int(input("Masukkan jumlah buku yang dibeli: "))

    # Operasi Aritmatika (Perkalian dan Pengurangan)
    subtotal = harga_buku * jumlah_buku
    pajak = 0.10 * subtotal  # Pajak 10%
    total_bayar = subtotal + pajak

    # Fungsi Output
    print("\n==================================")
    print("       NOTA PEMBAYARAN TOKO       ")
    print("==================================")
    print(f"Nama Pembeli : {nama_pembeli}")
    print(f"Subtotal     : Rp {subtotal:,.2f}")
    print(f"Pajak (10%)  : Rp {pajak:,.2f}")
    print(f"Total Bayar  : Rp {total_bayar:,.2f}")
    print("==================================")

hitung_pembayaran()
```
