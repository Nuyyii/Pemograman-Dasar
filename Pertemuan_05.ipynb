```python
# Data & Perhitungan
jumlah_buku = 7
harga_per_buku = 18500
diskon_persen = 0.10
uang_budi = 150000

total = jumlah_buku * harga_per_buku
diskon = total * diskon_persen
jumlah_dibayar = total - diskon
kembalian = uang_budi - jumlah_dibayar

# Menghitung Patungan
jumlah_orang = 4
iuran_per_orang = jumlah_dibayar // jumlah_orang
sisa = jumlah_dibayar % jumlah_orang

# Status Uang
apakah_cukup = uang_budi >= jumlah_dibayar

print("=" * 40)
print("         STRUK BELANJA BUKU   📚")
print("=" * 40)
print(f"Total Belanja    : Rp {int(total)}")
print(f"Diskon (10%)     : Rp {int(diskon)}")
print("-" * 40)
print(f"Total Harus Bayar: Rp {int(jumlah_dibayar)}")
print(f"Uang Kembalian   : Rp {int(kembalian)}")
print("-" * 40)
print("Patungan (4 Orang):")
print(f"   - Per Orang      : Rp {int(iuran_per_orang)}")
print(f"   - Sisa Uang      : Rp {int(sisa)}")
print("-" * 40)
status = "Cukup" if apakah_cukup else "Kurang"
print(f"STATUS UANG BUDI    : {status} (True)")
print("=" * 40)
```

```python
# Data panel surya
jumlah_panel = 12
daya_panel = 450
jam_penyinaran = 5
efisiensi = 0.8
kebutuhan_harian = 25
tarif_listrik = 1444.70

# Energi yang dihasilkan per-hari
energi_harian = (jumlah_panel * daya_panel / 1000) * jam_penyinaran * efisiensi

# energi per bulan dan nilai rupiah
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# kekurangan energi per-hari
kekurangan_energi = kebutuhan_harian - energi_harian

# Status PLTS
mencukupi = energi_harian >= kebutuhan_harian

# Menampilkan hasil
print('======== Pengeluaran Energi ========')
print(f"Energi harian  : {energi_harian} kWh")
print(f"Energi bulanan : {energi_bulanan} kWh")
print(f"Nilai listrik  : Rp {nilai_listrik}")
print("\n----- Status Energi -----")
print(f"Kekurangan     : {kekurangan_energi} kWh")
print(f"Mencukupi?     : {mencukupi}")
```

```python
# Data generator diesel
daya_generator = 200
jam_operasi = 8
konsumsi_bbm = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

# Energi yang dibangkitkan per hari
energi_harian = daya_generator * jam_operasi

# Kebutuhan solar dan biaya per hari
solar_harian = energi_harian * konsumsi_bbm
biaya_harian = solar_harian * harga_solar

# Waktu tangki mencukupi dan sisa solar
hari_cukup = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

# Rugi daya pada kabel
rugi_daya = arus ** 2 * hambatan

rugi_tinggi = rugi_daya > 150 and arus >= 20

# Menampilkan hasil
print('======== Generator Diesel ========')
print(f"Energi       : {energi_harian} kWh")
print(f"Solar        : {solar_harian} liter")
print(f"Biaya        : Rp {biaya_harian}")
print("\n--- Status Generator dan Kabel ---")
print(f"Tangki cukup : {hari_cukup} hari")
print(f"Sisa solar   : {sisa_solar} liter")
print(f"Rugi daya    : {rugi_daya} W")
print(f"Rugi tinggi? : {rugi_tinggi}")
```
