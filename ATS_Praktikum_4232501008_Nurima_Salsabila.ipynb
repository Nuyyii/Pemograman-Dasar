```python
# Nama        : Nurima Salsabila
# NIM         : 4232501008
# Tanggal     : 09 November 2026
# Mata Kuliah : Pemrograman Dasar

print("--- Soal 2: Energi harian dan Capacity Factor PLTD ---")
# Menyimpan nilai ke dalam variabel dan menghitung energi harian
daya_kw = 450
jam = 8
energi = daya_kw * jam

# Menyimpan variabel untuk kapasitas dan energi aktual
energi_aktual = 7200
kapasitas_kw = 400

# Menghitung capacity factor (%)
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Menampilkan hasil
print('\n\\\\\\', f"Energi harian = {energi} kWh", '///')
print('///', f"Capacity factor = {capacity_factor:.2f}%", '\\\\\\')
```

program ini untuk menghitung energi harian dan capacity factor pada PLTD. Pertama, program memberikan nilai pada variabel daya_kw = 450 kW dan waktu operasi selama jam = 8, kemudian mengalikan keduanya untuk memperoleh energi harian sebesar 3.600 kWh. Selanjutnya, program menghitung capacity factor berdasarkan energi_aktual = 7.200 kWh dibagi kapasitas_Kw = 400 kW yang dikalikan 24 jam, lalu dikalikan 100 untuk mendapatkan hasil dalam bentuk persentase. Terakhir, kedua hasil ditampilkan menggunakan print(), dengan format dua angka di belakang koma untuk capacity factor menggunakan (.2f).

```python
print("--- Soal 6: Status Operasi Boiler ---")
# Membaca input dan tekanan dari user
suhu = float(input('\nMasukkan Suhu (°C) : '))
tekanan = float(input('Masukkan Tekanan (bar) : '))

# Memeriksa kondisi secara berurutan
if suhu >= 550 or tekanan >= 14:
  print('STATUS : TRIP !')
elif suhu > 520 and tekanan > 12:
  print('STATUS : ! WARNING !')
else:
  print('STATUS : - NORMAL -')
```

Program ini digunakan untuk menentukan status kondisi operasional berdasarkan suhu dan tekanan. Pertama, program meminta pengguna memasukkan nilai suhu dan tekanan, kemudian mengubahnya menjadi tipe data float agar dapat membaca angka desimal. Selanjutnya, program memeriksa kondisi secara berurutan menggunakan percabangan if, elif, dan else. Jika suhu mencapai 550°C atau tekanan mencapai 14 bar, status yang ditampilkan adalah TRIP. Jika kondisi tersebut tidak terpenuhi, tetapi suhu lebih dari 520°C dan tekanan lebih dari 12 bar, statusnya adalah WARNING. Apabila kedua kondisi tersebut tidak terpenuhi, program menampilkan status NORMAL.

```python
print("--- Soal 8: Pengosongan Tangki Bahan Bakar ---")
# Mulai dengan volume 100
volume = 100

# Gunakan perulangan while
while volume > 0:
  volume -= 20
  print(f'\nVolume saat ini adalah {volume} liter')
  if volume == 20:
    print('\n>>> ! WARNING: Tangki Hampir Kosong ! <<<')


print('TANGKI KOSONG - Pengosongan Bahan Tangki Bakar Berhenti.')
```

Program ini digunakan untuk mensimulasikan proses pengosongan tangki bahan bakar menggunakan perulangan while. Pertama, variabel volume diatur sebesar 100 liter. Selanjutnya, perulangan akan terus berjalan selama volume masih lebih dari 0 liter. Pada setiap perulangan, volume dikurangi sebanyak 20 liter dan hasilnya ditampilkan. Ketika volume mencapai 20 liter, program menampilkan peringatan bahwa tangki hampir kosong. Setelah volume mencapai 0 liter, perulangan berhenti dan program menampilkan pesan bahwa tangki telah kosong sehingga proses pengosongan bahan bakar dihentikan

```python
print("--- Soal 10: Pembacaan Suhu dengan Sensor Rusak ---")
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Inisialisasi penghitung data tidak valid
data_tidak_valid = 0

# Periksa setiap data dengan perulangan 'for' dan nomor unit dimulai dari 1
for i, nilai_teks in enumerate(data, start=1):
    try:
        # Ubah data menjadi float
        nilai = float(nilai_teks)

        # b. Jika nilai melebihi 2.0, cetak Warning
        if nilai > 2.0:
            print(f"Warning: Unit {i} vibrasi tinggi!")

            # c. Jika nilai melebihi 3.0, cetak KRITIS dan hentikan perulangan dengan break
            if nilai > 3.0:
                print("KRITIS: pemeriksaan dihentikan")
                break

    except ValueError:
        # a. Jika gagal, cetak Unit [n]: data tidak valid
        print(f"\nUnit {i}: data tidak valid")
        # Tambahkan ke penghitung data tidak valid
        data_tidak_valid += 1
        # Lanjutkan ke unit berikutnya
        continue

# d. Setelah perulangan selesai, tampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {data_tidak_valid}")
```

Program ini digunakan untuk memeriksa data vibrasi pada setiap unit dan mendeteksi adanya data yang tidak valid maupun getaran yang terlalu tinggi. Pertama, program menyimpan data vibrasi dalam bentuk list dan menginisialisasi penghitung data tidak valid dengan nilai 0. Selanjutnya, perulangan for dengan fungsi enumerate() digunakan untuk memeriksa setiap data sekaligus memberikan nomor unit yang dimulai dari 1. Pada bagian try, data diubah menjadi tipe float. Jika nilai melebihi 2.0, program menampilkan peringatan (Warning), sedangkan jika melebihi 3.0, program menampilkan status KRITIS dan menghentikan perulangan menggunakan break. Apabila ditemukan data yang tidak dapat diubah menjadi angka, except ValueError akan menampilkan pesan bahwa data tidak valid, menambahkan penghitung sebanyak 1, dan menggunakan continue untuk melanjutkan pemeriksaan ke unit berikutnya. Terakhir, program menampilkan jumlah data yang tidak valid.

```python
print("--- Soal Bonus: Deteksi Anomali Suhu ---")
# Mengimpor modul random dan mendapatkan angka acak
import random
random.seed(42)

# Kondisi siklus awal 0 dan anomali True agar perulangan while dapat dilakukan
siklus = 0
ada_anomali = True

# Perulangan terus berlanjut selama ada anomali
while ada_anomali:
    siklus += 1
    ada_anomali = False
    print(f"\nSiklus ke-{siklus}:")
    # Menggunakan 6 sensor
    for sensor_id in range(1, 7):
        suhu = random.randint(65, 95)
        if suhu < 70 or suhu > 90:
            print(f"  ⚠ Sensor {sensor_id}: {suhu}°C — ANOMALI")
            ada_anomali = True
        else:
            print(f"  ✓ Sensor {sensor_id}: {suhu}°C — Normal")

print(f"\n~ Semua sensor normal pada siklus ke-{siklus}. ~")
```

Program ini digunakan untuk melakukan simulasi monitoring suhu pada enam sensor untuk mendeteksi adanya anomali. Pertama, program mengimpor modul random dan menggunakan random.seed(42) agar hasil angka acak tetap sama setiap program dijalankan. Selanjutnya, variabel siklus diatur mulai dari 0 dan ada_anomali diatur bernilai True agar perulangan while dapat dimulai. Pada setiap siklus, program memeriksa suhu dari enam sensor menggunakan perulangan for, dengan nilai suhu acak antara 65°C hingga 95°C. Jika suhu kurang dari 70°C atau lebih dari 90°C, sensor dinyatakan mengalami anomali dan program menampilkan peringatan. Jika suhu berada di antara 70°C hingga 90°C, sensor dinyatakan normal. Selama masih ditemukan anomali, pemeriksaan akan diulang hingga seluruh sensor berada dalam kondisi normal. Terakhir, program menampilkan informasi bahwa semua sensor telah normal beserta nomor siklus pemeriksaannya.
