```python
# Jawaban Latihan 2: Membuat Variabel sesuai dengan Tipe Data
nama_komponen = 'Resistor'
jumlah = 15
harga = 4500.12
tersedia = True

print ('Nama komponen : ', (nama_komponen), '/ Tipe :', type (nama_komponen))
print ('Jumlah        : ', (jumlah), '      / Tipe :', type (jumlah))
print ('Harga         : ', (harga), ' / Tipe :', type (harga))
print ('Tersedia      : ', (tersedia), '    / Tipe :', type (tersedia))

```

```python
# Jawaban Latihan 3: Menentukan Tipe Data
nama_unit = 'PLTS Nunukan'
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = 'benar'

print (nama_unit, '>>>', type(nama_unit))
print (jumlah_panel, '         >>>', type (jumlah_panel))
print (tegangan_keluaran, '        >>>', type(tegangan_keluaran))
print (sedang_beroperasi, '       >>>', type(sedang_beroperasi))
```

```python
# Jawaban Latihan 4: Tegangan pada sebuah Panel
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

#(a) cetak pembacaan pertama dan terakhir
print ('Pembacaan pertama :', tegangan_harian[0], '|',
       'Pembacaan terakhir :', tegangan_harian[5])

#(b) menambahkan satu pembacaan baru dengan append
tegangan_harian.append (222.0)

print (tegangan_harian)

#(c) jumlah total data
print ('Jumlah total data : ', len(tegangan_harian))

```

```python
# Jawaban Latihan 5: Spesifikasi motor listrik
motor = {
    'nama': "Motor Induksi",
    'daya_kW' : 5.8,
    'tegangan_V' : 338,
    'rpm' : 3600
}

# mencetak salah satu nilai motor
print ('Daya : ', motor ['daya_kW'], 'kW')

# mengubah nilai rpm
motor ['rpm'] = 2800

print (motor)
```
