```python
# >>> AKTIVITAS 6A <<<
suhu = [28, 31, 30, 33, 29]       # suhu (°C) pada 5 hari

print("suhu[0] =", suhu[0])       # elemen pertama
print("suhu[4] =", suhu[4])       # elemen terakhir (len - 1)
print("suhu[-1] =", suhu[-1])     # indeks negatif: dihitung dari belakang

# --- tebak dulu, baru jalankan ---
print(suhu[1])                 # hasil tebakan >> 31
print(suhu[-2])                # hasil tebakan >> 33
print(suhu[len(suhu) - 1])     # hasil tebakan >> 29
print(suhu[2] + suhu[3])       # hasil tebakan >> 63
```

```python
# >>> AKTIVITAS 6F <<<
skor = [80, 65, 90, 75, 70]

total = 0                    # akumulator: mulai dari nol
jumlah_lulus = 0             # penghitung = akumulator yang menambah 1
for nilai in skor:
    total = total + nilai    # tambahkan setiap elemen
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
# >>> AKTIVITAS 6G <<<
suhu = [28, 31, 30, 33, 29, 27, 32]

# Rata-rata bergerak 5 hari: jendela = 5 nilai yang berakhir di indeks i
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

```python
# >>> AKTIVITAS 6H <<<
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# Perubahan harian: mulai dari indeks 1 karena dibandingkan dengan elemen sebelumnya
for i in range(1, len(energi)):
    perubahan = (energi[i] - energi[i - 1]) / energi[i - 1] * 100
    if perubahan < -20:
        print(f"Hari {i + 1}: {perubahan:+.1f}% dibanding hari sebelumnya, perlu diselidiki")

# Rata-rata bergerak 7 hari
rata_7 = []
for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7.append(sum(jendela) / 7)

print(f"{len(rata_7)} nilai rata-rata bergerak (data {len(energi)} hari)")
print(f"Terendah {min(rata_7):,.0f} MWh/hari, tertinggi {max(rata_7):,.0f} MWh/hari")
```

```python
# >>> AKTIVITAS 6I <<<
suhu = [28, 31, 29, 34, 30, 27, 32]
print("List suhu awal        :", suhu)

# Tampilkan 3 hari pertama dan 2 hari terakhir dengan slicing
tiga_pertama = suhu[:3]
dua_terakhir = suhu[-2:]
print("3 hari pertama        :", tiga_pertama)
print("2 hari terakhir       :", dua_terakhir)

# Tambahkan satu hari dengan append, lalu keluarkan lagi dengan pop
suhu.append(33)          # Menambahkan suhu hari baru (misal: 33) di akhir
print("Setelah append(33)    :", suhu)

suhu_keluar = suhu.pop() # Mengeluarkan elemen terakhir kembali
print("Suhu yang di-pop      :", suhu_keluar)
print("Setelah pop           :", suhu)

# Tampilkan versi urutnya dengan sorted tanpa mengubah list asli
suhu_urut = sorted(suhu)
print("Versi terurut (sorted):", suhu_urut)
print("List asli tetap       :", suhu)  # List asli tidak berubah urutannya

# Hitung rata-rata dan cari suhu tertinggi beserta hari ke berapa
total_suhu = sum(suhu)
rata_rata = total_suhu / len(suhu)
suhu_tertinggi = max(suhu)
hari_tertinggi = suhu.index(suhu_tertinggi) + 1  # Ditambah 1 karena indeks mulai dari 0

print(f"Rata-rata suhu        : {rata_rata:.1f}°C")
print(f"Suhu tertinggi        : {suhu_tertinggi}°C (pada hari ke-{hari_tertinggi})")

# Buat list baru berisi suhu di atas rata-rata
suhu_di_atas_rata = [s for s in suhu if s > rata_rata]
print("Suhu di atas rata-rata:", suhu_di_atas_rata)
```

```python
# >>> AKTIVITAS 6J <<<
nilai = [
    [80, 70, 90],     # baris 0: Ana
    [65, 75, 60],     # baris 1: Budi
    [90, 85, 95],     # baris 2: Cici
]

print(nilai[1])        # satu baris utuh
print(nilai[1][2])     # baris 1, kolom 2: ujian 3 milik Budi
print("baris:", len(nilai), " kolom:", len(nilai[0]))

# Total untuk setiap baris (setiap mahasiswa)
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Satu kolom: indeks yang sama di setiap baris (ujian 1 semua orang)
ujian1 = []
for baris in nilai:
    ujian1.append(baris[0])
print("Skor ujian 1:", ujian1)
print("Skor ujian 2 Cici:", nilai[2][1])  # Hasilnya: 85
print("Jumlah ujian (kolom):", len(nilai[0])) # Hasilnya: 3
```

```python
# Pemanasan: Lima latihan kecil (data kecil)
skor = [80, 65, 90, 75, 70, 85]

# (1) Mencetak skor pertama dan skor terakhir (pakai indeks negatif)
print("Skor pertama :", skor[0])
print("Skor terakhir:", skor[-1])

# (2) Mencetak rata-rata
rata_rata = sum(skor) / len(skor)
print(f"Rata-rata    : {rata_rata:.1f}")

# (3) Menghitung berapa skor di atas rata-rata
di_atas_rata = [s for s in skor if s > rata_rata]
print("Jumlah skor di atas rata-rata:", len(di_atas_rata))

# (4) Mencetak posisi skor terendah sebagai nomor mahasiswa (mahasiswa pertama = 1)
skor_terendah = min(skor)
indeks_terendah = skor.index(skor_terendah)
nomor_mahasiswa = indeks_terendah + 1
print(f"Skor terendah adalah {skor_terendah}, milik Mahasiswa ke-{nomor_mahasiswa}")

# (5) Membuat list baru dengan setiap skor ditambah 5
skor_plus_lima = [s + 5 for s in skor]
print("Skor baru (+5):", skor_plus_lima)
```

```python
# Latihan Praktikum: Laporan Produksi Bulanan
# Data energi 30 hari (disalin dari Aktivitas 6H)
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

print("=== LAPORAN PRODUKSI ENERGI BULANAN ===\n")

# 1. Total dan rata-rata energi
total_energi = sum(energi)
rata_bulanan = total_energi / len(energi)
print(f"1. Total Energi     : {total_energi:,} MWh")
print(f"   Rata-rata Harian : {rata_bulanan:,.2f} MWh\n")

# 2. Tiga hari tertinggi dan tiga hari terendah (nomor hari dan nilainya)
# Membuat pasangan (nilai, nomor_hari) agar nomor hari tidak hilang saat diurutkan
data_berpasangan = [(energi[i], i + 1) for i in range(len(energi))]
data_urut = sorted(data_berpasangan)

tiga_terendah = data_urut[:3]
tiga_tertinggi = data_urut[-3:][::-1] # Diambil 3 terakhir lalu dibalik agar urut dari terbesar

print("2. Tiga Hari Tertinggi:")
for nilai, hari in tiga_tertinggi:
    print(f"   - Hari ke-{hari:2d}: {nilai:,} MWh")

print("   Tiga Hari Terendah:")
for nilai, hari in tiga_terendah:
    print(f"   - Hari ke-{hari:2d}: {nilai:,} MWh")
print()

# 3. Jumlah hari di bawah target yang dimasukkan
target = float(input("3. Masukkan target produksi harian (MWh, misal 6000): "))
hari_bawah_target = [e for e in energi if e < target]
print(f"   Jumlah hari di bawah target ({target:,.0f} MWh): {len(hari_bawah_target)} hari\n")

# 4. Setiap hari yang rata-rata bergerak 7 harinya turun di bawah 97% rata-rata bulanan
batas_97 = 0.97 * rata_bulanan
print(f"4. Hari dengan Rata-rata Bergerak 7 Hari < 97% rata-rata bulanan (< {batas_97:,.2f} MWh):")
ada_anomali = False
for i in range(6, len(energi)):
    jendela_7 = energi[i - 6 : i + 1]
    rata_7 = sum(jendela_7) / 7
    if rata_7 < batas_97:
        print(f"   - Hari ke-{i + 1:2d}: rata-rata 7 hari = {rata_7:,.2f} MWh")
        ada_anomali = True

if not ada_anomali:
    print("   - Tidak ada hari yang turun di bawah batas 97%.")
print()

# 5. Total energi per minggu (minggu 1–4, ditambah sisa hari)
print("5. Total Energi per Minggu:")
for minggu in range(4):
    awal = minggu * 7
    akhir = awal + 7
    total_minggu = sum(energi[awal:akhir])
    print(f"   - Minggu ke-{minggu + 1} (Hari {awal + 1:2d}–{akhir:2d}): {total_minggu:,} MWh")

# Menghitung sisa hari (indeks 28 ke atas, yaitu hari ke-29 dan ke-30)
sisa_hari = energi[28:]
if sisa_hari:
    print(f"   - Sisa hari    (Hari 29–{len(energi):2d}): {sum(sisa_hari):,} MWh")
```
