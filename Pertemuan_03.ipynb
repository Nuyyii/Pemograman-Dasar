```python
#Nama    : Nurima Salsabila
#NIM     : 4232501008
#Tanggal : 24 September 2026
#Tujuan  : Menghitung efisiensi termal suatu unit
#          pembangkit berdasarkan energi input dan output

# Input energi input dan energi output (dalam MJ)
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

# Menghitung efisiensi termal dengan tanda kurung yang sesuai operator precedence
efisiensi = (energi_output / energi_input) * 100

# Menampilkan hasil dengan f-string dan pembulatan 2 angka desimal (.2f)
print(f"Efisiensi termal pembangkit adalah {efisiensi:.2f}%")

```

```python
#Nama    : Nurima Salsabila
#NIM     : 4232501008
#Tanggal : 24 September 2026
#Tujuan  : Menentukan suatu unit pembangkit memerlukan
#          maintenance berdasarkan suhu boiler dan beban operasi

# Input suhu boiler dan beban operasi
suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

# Penilaian kondisi maintenance menggunakan operator relasional dan logika (or)
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

# Menampilkan hasil dalam kalimat naratif menggunakan f-string
print(f"Status kondisi unit perlu maintenance: {perlu_maintenance}")

```
